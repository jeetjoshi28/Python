Understanding Expections

Expection handling in python allows you to handel errors gracefully and take corrective actions without stopping the execution of the program.

What are the Expections?

Expections are events that disrupt the normal flow a program. They occur when an error is encountered during program execution. Common expection. Common expections include:

    ZeroDivisionError: Dividing bt zero
    FileNotFoundError: File not found
    ValueError: Invalid value
    TypeError: Invalid type

```python
## Expection try, expection, block
try:
    a=b
except NameError as ex:
    print(ex)
```

```python
try:    
    result = 10/0
except ZeroDivisionError as ex:
    print(ex)
    print("You cannot divide by zero")
```

```python
try:    
    result = 1/2
    a=b
except ZeroDivisionError as ex:
    print(ex)
    print("You cannot divide by zero")
except Exception as ex1:
    print(ex1)
    print("Main expectin block")
```

```python
try:
    num=int(input("Enter a number: "))
    result = 10/num
    print(result)
except ValueError:
    print("This is not a valid number") #  pass the string
except ZeroDivisionError as ex:
    print(ex)
    print("You cannot divide by zero") # pass the 0 
```

```python
#try except and finally
try:
    num=int(input("Enter a number: "))
    result = 10/num
    print(result)
except ValueError:
    print("This is not a valid number") #  pass the string
except ZeroDivisionError as ex:
    print(ex)
    print("You cannot divide by zero") # pass the 0 
except Exception as ex:
    print(ex)
    print("Main exception block")
finally:
    print("This will always execute")
```
